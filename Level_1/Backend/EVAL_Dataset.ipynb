```python
from deepeval.synthesizer import Synthesizer
from deepeval.synthesizer.config import ContextConstructionConfig
from deepeval.models import OllamaModel, OllamaEmbeddingModel

llm = OllamaModel(
    model="qwen2.5:7b-instruct-q4_K_M",
    base_url="http://localhost:11434",
)

embedder = OllamaEmbeddingModel(
    model="nomic-embed-text",
    base_url="http://localhost:11434",
)

config = ContextConstructionConfig(
    embedder=embedder,
    critic_model=llm,
)

```

```python
synthesizer = Synthesizer(
    model=llm,
    async_mode=False,
)


file_path = r"/home/rushikesh/Selvin/Level_1/example.txt"

goldens = synthesizer.generate_goldens_from_docs(
    document_paths=[file_path],
    context_construction_config=config,
    max_goldens_per_context=10,
)

print(f"Generated {len(goldens)} goldens")
```

```python
for i, golden in enumerate(goldens, 1):
    print(f"\n{'=' * 80}")
    print(f"GOLDEN {i}")
    print(f"{'=' * 80}")

    print("\nINPUT:")
    print(golden.input)

    print("\nEXPECTED OUTPUT:")
    print(golden.expected_output)

    print("\nCONTEXT:")
    for context in golden.context:
        print(f"  - {context}")

    print("\nEVOLUTION:")
    print(golden.additional_metadata.get("evolutions"))

    print("\nQUALITY:")
    print(f"  Input quality:   {golden.additional_metadata.get('synthetic_input_quality')}")
    print(f"  Context quality: {golden.additional_metadata.get('context_quality')}")

    print("\nSOURCE:")
    print(golden.source_file)
```
